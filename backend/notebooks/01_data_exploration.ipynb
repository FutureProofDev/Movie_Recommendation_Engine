```python
import pandas as pd

# Load ratings
ratings = pd.read_csv('../../data/ml-100k/u.data', sep='\t',
                       names=['user_id', 'item_id', 'rating', 'timestamp'])

# Load user info
users = pd.read_csv('../../data/ml-100k/u.user', sep='|',
                     names=['user_id', 'age', 'gender', 'occupation', 'zip_code'])

# Load movie info 
genre_cols = ['unknown', 'Action', 'Adventure', 'Animation', "Children's", 'Comedy',
              'Crime', 'Documentary', 'Drama', 'Fantasy', 'Film-Noir', 'Horror',
              'Musical', 'Mystery', 'Romance', 'Sci-Fi', 'Thriller', 'War', 'Western']
movies = pd.read_csv('../../data/ml-100k/u.item', sep='|', encoding='latin-1',
                      names=['item_id', 'title', 'release_date', 'video_release_date',
                             'IMDb_URL'] + genre_cols)

print("Ratings shape:", ratings.shape)
print("Users shape:", users.shape)
print("Movies shape:", movies.shape)

ratings.head()
```

```python
# Check for missing values
print("Ratings missing:\n", ratings.isnull().sum())
print("\nUsers missing:\n", users.isnull().sum())
print("\nMovies missing:\n", movies.isnull().sum())

# Check for duplicates
print("\nDuplicate ratings:", ratings.duplicated().sum())

# Basic stats
print("\nRating distribution:\n", ratings['rating'].value_counts().sort_index())
print("\nRatings per user (avg):", ratings.groupby('user_id').size().mean())
print("Ratings per movie (avg):", ratings.groupby('item_id').size().mean())
```

```python
movies_clean = movies.drop(columns = ['release_date', 'video_release_date', 'IMDb_URL'])

ratings_movies = ratings.merge(movies_clean, on='item_id')

full_df = ratings_movies.merge(users, on='user_id')

print(full_df.shape)
full_df.head()
```

```python
user_item_matrix = ratings.pivot_table(index='user_id', columns='item_id', values='rating')
print(user_item_matrix.shape)
user_item_matrix.head()
```
